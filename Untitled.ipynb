The spectral resolution of a Fabry–Pérot interferometer is determined by its finesse and free spectral range (FSR). The finesse describes how narrow the transmission peaks are relative to the spacing between neighbouring resonances. For mirrors with identical intensity reflectivity \(R\), the theoretical finesse is given by

$$
\mathcal{F}=\frac{\pi\sqrt{R}}{1-R}.
$$

A higher mirror reflectivity therefore leads to a higher finesse and narrower transmission peaks. The corresponding minimum resolvable frequency separation, according to the Rayleigh criterion, is given by the full width at half maximum (FWHM) of a transmission peak:

$$
\Delta\nu_{\min}=\Gamma_{\mathrm{FWHM}}=\frac{\mathrm{FSR}}{\mathcal{F}}.
$$

Consequently, a higher finesse results in a smaller minimum resolvable frequency separation and thus improves the spectral resolution of the interferometer. In this experiment, the finesse is determined theoretically from the mirror reflectivity. The Lorentzian fits to the intensity profiles recorded by the CMOS camera describe the spatial widths of the observed interference rings. These widths can be used to characterise the sharpness of the rings, but they do not directly represent the spectral linewidth of the transmission resonances.
